```python
import librosa
import soundfile
import numpy as np
import os, glob, pickle
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
```

```python
def add_noise(data, noise_factor=0.005):
    noise = np.random.randn(len(data))
    augmented = data + noise_factor * noise
    return augmented
```

```python
def extract_feature(file_name, mfcc, chroma, mel):
    X, sample_rate = librosa.load(
        file_name,
        sr=22050,
        res_type='kaiser_fast',
        duration=3.0,      # FIXED length
        offset=0.5         # skip silence at start
    )
    X = librosa.util.normalize(X)

    # add noise augmentation
    #if np.random.rand() > 0.85:
     #   X = add_noise(X)

    result = np.array([])

    if mfcc:
        mfcc = librosa.feature.mfcc(y=X, sr=sample_rate, n_mfcc=40)

        delta = librosa.feature.delta(mfcc)
        delta2 = librosa.feature.delta(mfcc, order=2)

        mfcc_mean = np.mean(mfcc.T, axis=0)
        mfcc_std = np.std(mfcc.T, axis=0)

        delta_mean = np.mean(delta.T, axis=0)
        delta2_mean = np.mean(delta2.T, axis=0)
        
        result = np.hstack((mfcc_mean, mfcc_std, delta_mean, delta2_mean))

    if chroma:
        stft = np.abs(librosa.stft(X))
        chroma = librosa.feature.chroma_stft(S=stft, sr=sample_rate)

        chroma_mean = np.mean(chroma.T, axis=0)
        chroma_std = np.std(chroma.T, axis=0)

        result = np.hstack((result, chroma_mean, chroma_std))

    if mel:
        mel = librosa.feature.melspectrogram(y=X, sr=sample_rate)
        mel = librosa.power_to_db(mel, ref=np.max)

        mel_mean = np.mean(mel.T, axis=0)
        mel_std = np.std(mel.T, axis=0)

        result = np.hstack((result, mel_mean, mel_std))
    pitch, _ = librosa.piptrack(y=X, sr=sample_rate)

    pitch_mean = np.mean(pitch)
    pitch_std = np.std(pitch)

    result = np.hstack((result, pitch_mean, pitch_std))

    return result
```

```python
# emotions
emotions={
  '01':'neutral',
  '02':'calm',
  '03':'happy',
  '04':'sad',
  '05':'angry',
  '06':'fearful',
  '07':'disgust',
  '08':'surprised'
}


observed_emotions=['happy','sad','angry','fearful']  #only these 4 are used for training
```

```python
def load_data(test_size=0.2):
    x, y = [], []

    # Since Actor_* folders are in the SAME directory as the notebook
    # 1. load RAVDESS dataset
    for file in glob.glob("Actor_*/*.wav"):
        file_name = os.path.basename(file)

        # Extract emotion code from filename
        emotion = emotions[file_name.split("-")[2]]

        # Filter only required emotions
        if emotion not in observed_emotions:
            continue

        # Extract audio features
        feature = extract_feature(file, mfcc=True, chroma=True, mel=True)

        x.append(feature)
        y.append(emotion)

    
    # 2. load TESS dataset
    for file in glob.glob("TESS/*/*.wav"):
        
        file_name = os.path.basename(file)
    
        # emotion is the last word in filename
        emotion = file_name.split("_")[-1]
        if emotion == "fear":
            emotion = "fearful"
    
        if emotion not in observed_emotions:
            continue
    
        feature = extract_feature(file, mfcc=True, chroma=True, mel=True)
    
        x.append(feature)
        y.append(emotion)


    return train_test_split(
        np.array(x),
        y,
        test_size=test_size,
        random_state=9,
        stratify = y
    )
```

```python
#DataFlair - Split the dataset
x_train,x_test,y_train,y_test=load_data(test_size=0.25)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#DataFlair - Get the shape of the training and testing datasets
print((x_train.shape[0], x_test.shape[0]))
```

```python
#DataFlair - Get the number of features extracted
print(f'Features extracted: {x_train.shape[1]}')
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score

model = SVC(
    kernel='rbf',
    C=50,
    gamma='scale',
    class_weight='balanced',
    probability=True
)

# 5-fold cross validation
scores = cross_val_score(model, x_train, y_train, cv=5)
```

```python
#DataFlair - Train the model
model.fit(x_train,y_train)
```

```python
#DataFlair - Predict for the test set
y_pred=model.predict(x_test)
```

```python
#DataFlair - Calculate the accuracy of our model
accuracy=accuracy_score(y_true=y_test, y_pred=y_pred)

#DataFlair - Print the accuracy
print("Accuracy: {:.2f}%".format(accuracy*100))
```

```python
def predict_long_audio(file_path):

    audio, sr = librosa.load(file_path, sr=22050)

    window = 3 * sr
    predictions = []

    for start in range(0, len(audio), window):

        segment = audio[start:start+window]

        if len(segment) < window:
            continue

        # save temporary segment
        import soundfile as sf
        sf.write("temp.wav", segment, sr)

        feature = extract_feature("temp.wav", mfcc=True, chroma=True, mel=True)

        feature = feature.reshape(1, -1)

        feature = scaler.transform(feature)

        pred = model.predict(feature)[0]

        predictions.append(pred)

    return max(set(predictions), key=predictions.count)
```

```python
test_audio = "test.wav"

emotion = predict_long_audio(test_audio)

print("Predicted Emotion:", emotion)
```
